```python
G.add_edges_from(Edges)
color=[]
colors=['yellow','red','blue','pink','magenta','green','violet','brown','grey','orange']
for node in G.nodes():
    #print(len(df['Line'][df['Station name']==node]))
    linenum=list(df['Line'][df['Station name']==node])
    if len(linenum)==1:
        #print(linenum[0])
        color.append(colors[linenum[0]-1])
    else:
        color.append('black')
```

```python
counts=0
mapping=dict()
for node in G.nodes():
    mapping[node]=counts
    counts=counts+1
```

```python
Gc=nx.relabel_nodes(G,mapping,copy=True)
```

```python
fig=plt.figure(figsize=(12,12),dpi=200)
neato=nx.nx_agraph.graphviz_layout(Gc,prog='neato')
kk=nx.kamada_kawai_layout(Gc)
nx.draw(Gc,with_labels=True,pos=neato,
        font_size=8,node_color=color,node_size=50)
plt.show()
```
